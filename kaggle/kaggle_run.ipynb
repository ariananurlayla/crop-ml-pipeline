```python
import glob, os, shutil

# locate the two attached datasets
csvs = sorted(glob.glob('/kaggle/input/*/IFRI_181_Band_Dataset_Extract.csv'))
assert csvs, 'CSV dataset not found - attach the dataset containing IFRI_181_Band_Dataset_Extract.csv'
csv_path = csvs[0]

pipes = glob.glob('/kaggle/input/*/main.py')
assert pipes, 'pipeline dataset not found - attach a dataset zipped from the crop-ml-pipeline folder'
src, dst = os.path.dirname(pipes[0]), '/kaggle/working/crop-ml-pipeline'
shutil.copytree(src, dst)

print('csv   :', csv_path)
print('pipeline copied to', dst)
```

```python
import shlex

%cd /kaggle/working/crop-ml-pipeline
!python main.py --data {shlex.quote(csv_path)} --task both --binary --attention --chain \
    --epochs 150 --batch-size 128 --device cuda --outputs /kaggle/working/outputs
```
